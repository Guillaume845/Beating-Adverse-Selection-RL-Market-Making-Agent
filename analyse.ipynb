```python
import matplotlib.pyplot as plt
import numpy as np
from stable_baselines3 import PPO
from market_env import GlostenMilgromEnv

env = GlostenMilgromEnv(mu=0.3, gamma=0.05)
model = PPO.load("ppo_market_maker")

obs, _ = env.reset()
inventories = []
spreads = []

done = False
truncated = False

while not (done or truncated):
    action, _ = model.predict(obs, deterministic=True)
    obs, _, done, truncated, _ = env.step(action)
    
    inventories.append(obs[0])
    spreads.append(action + 1)

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 8), sharex=True)

ax1.plot(inventories, color='#1f77b4', drawstyle='steps-post', linewidth=2)
ax1.axhline(0, color='black', linestyle='--', alpha=0.3)
ax1.set_ylabel('Position d\'inventaire')
ax1.set_title('Comportement de l\'Agent Market Maker')

ax2.plot(spreads, color='#ff7f0e', drawstyle='steps-post', linewidth=2)
ax2.set_ylabel('Spread coté (Ticks)')
ax2.set_xlabel('Temps')
ax2.set_yticks([1, 2, 3, 4, 5])

plt.tight_layout()
plt.show()
```
